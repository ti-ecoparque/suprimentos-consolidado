```python
import os
import pandas as pd

# 1. Define o caminho correto usando o nome exato da pasta visto na imagem
diretorio_atual = os.getcwd()
pasta_excel = os.path.join(diretorio_atual, "xls_solicitacao")

if not os.path.exists(pasta_excel):
    raise FileNotFoundError(f"⚠️ A pasta não foi encontrada no caminho: {pasta_excel}")

# 2. Modificado para aceitar tanto .xlsx quanto .xls (usando endswith)
arquivos_excel = [
    f for f in os.listdir(pasta_excel) 
    if f.lower().endswith(".xlsx") or f.lower().endswith(".xls")
]

if not arquivos_excel:
    raise FileNotFoundError("⚠️ Nenhum arquivo Excel (.xls ou .xlsx) foi encontrado na pasta!")

# Pega o primeiro arquivo encontrado
nome_arquivo = arquivos_excel[0]
caminho_completo = os.path.join(pasta_excel, nome_arquivo)

print(f"📂 Arquivo detectado com sucesso: {nome_arquivo}")

# 3. Lê o Excel para o DataFrame
df = pd.read_excel(caminho_completo)

# Configura o Pandas para exibir TODAS as colunas na tela
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)

# Mostra as colunas e as primeiras 5 linhas
df.head()
```
